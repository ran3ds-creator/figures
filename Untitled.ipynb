```python
import numpy as np
import matplotlib.pyplot as plt

# Use the same converted, background-corrected rates as Figure 2.
data = np.loadtxt('Figure2_summary.csv', delimiter=',', skiprows=1)
substrate = data[:, 0]       # mM
rate = data[:, 4]            # Mean NADH consumption rate, µM/s

# Calculate the two linear transformations.
inverse_s = 1 / substrate
inverse_v = 1 / rate
s_over_v = substrate / rate

table = np.column_stack((substrate, rate, inverse_s, inverse_v, s_over_v))
np.savetxt('Figure3_transformed_data.csv', table, delimiter=',',
           header='Pyruvate_mM,Mean_rate_uM_s,Inverse_S_per_mM,Inverse_v_s_per_uM,S_over_v_mM_s_per_uM',
           comments='', fmt='%.9f')

plots = [
    (inverse_s, inverse_v, 'Lineweaver–Burk',
     '1/[Pyruvate] (mM⁻¹)', '1/v (s/µM)', 'Lineweaver_Burk'),
    (substrate, s_over_v, 'Hanes–Woolf',
     '[Pyruvate] (mM)', '[Pyruvate]/v (mM·s/µM)', 'Hanes_Woolf')
]

combined, axes = plt.subplots(1, 2, figsize=(10, 4))
results = []

for i, (x, y, title, xlabel, ylabel, filename) in enumerate(plots):
    # Fit all ten transformed means without removing any points.
    slope, intercept = np.polyfit(x, y, 1)
    predicted = slope * x + intercept
    r_squared = 1 - np.sum((y - predicted)**2) / np.sum((y - y.mean())**2)

    if i == 0:
        vmax = 1 / intercept
        km = slope / intercept
    else:
        vmax = 1 / slope
        km = intercept / slope

    results.append([slope, intercept, r_squared, vmax, km])
    print(f'{title}: R² = {r_squared:.4f}')
    print(f'Calculated Vmax = {vmax:.4f} µM/s; Km = {km:.4f} mM')
    if vmax <= 0 or km <= 0:
        print('Nonphysical estimates: do not report these as valid enzyme constants.')

    line_x = np.linspace(0, max(x) * 1.05, 200)
    fig, ax = plt.subplots(figsize=(5.2, 4.2))

    # Save each plot separately and also in a combined figure.
    for current_ax in [ax, axes[i]]:
        current_ax.scatter(x, y, color='#0072B2', s=28,
                           label='Transformed mean rates')
        current_ax.plot(line_x, slope * line_x + intercept,
                        color='#D55E00', linewidth=1.2,
                        label=f'Linear fit ($R^2$ = {r_squared:.3f})')
        current_ax.axhline(0, color='0.7', linewidth=0.7)
        current_ax.set_xlabel(xlabel)
        current_ax.set_ylabel(ylabel)
        current_ax.set_title(title)
        current_ax.spines['top'].set_visible(False)
        current_ax.spines['right'].set_visible(False)
        current_ax.legend(frameon=False, fontsize=8)

    fig.tight_layout()
    fig.savefig(f'Figure3_{filename}.png', dpi=600, bbox_inches='tight')
    fig.savefig(f'Figure3_{filename}.svg', bbox_inches='tight')
    plt.close(fig)

combined.tight_layout()
combined.savefig('Figure3_combined.png', dpi=600, bbox_inches='tight')
combined.savefig('Figure3_combined.svg', bbox_inches='tight')

np.savetxt('Figure3_fit_results.csv', results, delimiter=',',
           header='Slope,Intercept,R_squared,Calculated_Vmax_uM_s,Calculated_Km_mM',
           comments='', fmt='%.9f')
# Fit-results row 1 = Lineweaver-Burk; row 2 = Hanes-Woolf.
plt.show()
```
