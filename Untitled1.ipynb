```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# Read CONVERTED Excel data: time, followed by 18 individual NADH columns.
# Keep the CSV beside this script, or in the notebook's working folder.
folder = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
data = np.loadtxt(folder / "Figure1_NADH_uM.csv", delimiter=",", skiprows=1)
if data.shape != (10, 19):
    raise ValueError("Expected 10 time rows and 19 columns: time plus 18 trials.")
if not np.isfinite(data).all():
    raise ValueError("The CSV contains a missing or nonnumeric value.")
time = data[:, 0]
if not np.array_equal(time, np.arange(0, 91, 10)):
    raise ValueError("Time must run from 0 to 90 seconds in 10-second steps.")

# Each group contains the three INDIVIDUAL trials for that LDH level.
groups = {
    "0":    data[:, 1:4],    # Excel columns B-D
    "0.06": data[:, 4:7],    # Excel columns E-G
    "0.12": data[:, 7:10],   # Excel columns H-J
    "0.18": data[:, 10:13],  # Excel columns K-M
    "0.37": data[:, 13:16],  # Excel columns N-P
    "0.56": data[:, 16:19],  # Excel columns Q-S
}

plt.rcParams.update({
    "font.family": "DejaVu Sans", "font.size": 9,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.linewidth": 0.8, "svg.fonttype": "none",
})
colors = ["#0072B2", "#D55E00", "#009E73"]
markers = ["o", "s", "^"]
fig, axes = plt.subplots(2, 3, figsize=(9.2, 5.6), sharex=True, sharey=True)

for panel, (concentration, trials) in enumerate(groups.items()):
    ax = axes.flat[panel]
    r_squared = []
    for trial in range(3):
        measured = trials[:, trial]
        ax.plot(time, measured, color=colors[trial], marker=markers[trial],
                markersize=3.4, linewidth=0.9, markeredgewidth=0.5,
                markerfacecolor="white")

        # Fit each trial separately over the full 0-90 s recording.
        slope, intercept = np.polyfit(time, measured, 1)
        predicted = slope * time + intercept
        total = np.sum((measured - measured.mean())**2)
        r_squared.append(1 - np.sum((measured - predicted)**2) / total)

        # Show dashed fits only for the selected condition, panel C.
        if concentration == "0.12":
            ax.plot(time, predicted, "--", color=colors[trial],
                    linewidth=1.0, alpha=0.75)

    ax.set_title(f"{concentration} U/mL", fontsize=10, pad=7)
    ax.text(0.02, 1.03, "ABCDEF"[panel], transform=ax.transAxes,
            fontsize=11, fontweight="bold", va="bottom")
    ax.text(0.97, 0.96,
            f"$R^2$: {min(r_squared):.3f}–{max(r_squared):.3f}",
            transform=ax.transAxes, ha="right", va="top", fontsize=8)
    ax.set(xlim=(-2, 92), ylim=(0, 135))
    ax.set_xticks([0, 30, 60, 90])
    ax.set_yticks([0, 30, 60, 90, 120])
    ax.tick_params(direction="out", length=3)
    if panel >= 3:
        ax.set_xlabel("Time (s)")
    if panel % 3 == 0:
        ax.set_ylabel("[NADH] (µM)")

handles = [Line2D([0], [0], color=colors[i], marker=markers[i],
                  markerfacecolor="white", linewidth=0.9, markersize=4,
                  label=f"Recorded trial {i + 1}") for i in range(3)]
handles.append(Line2D([0], [0], color="0.35", linestyle="--",
                      label="Linear fits in panel C"))
fig.legend(handles=handles, loc="lower center", ncol=4, frameon=False,
           fontsize=8, bbox_to_anchor=(0.5, 0.01))
fig.subplots_adjust(left=0.08, right=0.99, top=0.92, bottom=0.16,
                    hspace=0.40, wspace=0.15)

# PNG for insertion into the report; SVG for a sharp vector original.
fig.savefig(folder / "Figure1_progress_curves.png", dpi=600,
            bbox_inches="tight", facecolor="white")
fig.savefig(folder / "Figure1_progress_curves.svg",
            bbox_inches="tight", facecolor="white")
plt.show()
```
